````python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp


# ---------------------------------------------------------
# 1. CREATE THE TIME VALUES
# ---------------------------------------------------------

# Create 1,000 evenly spaced time values from 0 to 40 minutes.
# These values will be used as the x-axis of the graph.
time = np.linspace(0, 40, 1000)


# ---------------------------------------------------------
# 2. DEFINE THE DISTANCE EQUATIONS
# ---------------------------------------------------------

# The robber travels at 2.5 km/min.
#
# Mathematical equation:
#     d_r = 2.5t

robber_distance = 2.5 * time


# The sheriff starts 5 minutes later and travels at 3 km/min.
#
# Mathematical equation:
#     d_s = 3(t - 5)

sheriff_distance = 3 * (time - 5)


# ---------------------------------------------------------
# 3. SOLVE THE EQUATION AUTOMATICALLY
# ---------------------------------------------------------

# Create a symbolic variable called t.
t = sp.symbols('t')

# Define the equation:
#
#     2.5t = 3(t - 5)
#
# The equation represents the moment when the robber
# and sheriff have traveled the same distance.

equation = sp.Eq(2.5 * t, 3 * (t - 5))

# Ask SymPy to solve the equation for t.
catch_time = sp.solve(equation, t)[0]

# Convert the SymPy result to a regular Python float.
catch_time = float(catch_time)

# Calculate the distance at the capture point.
catch_distance = 2.5 * catch_time


# ---------------------------------------------------------
# 4. DISPLAY THE SOLUTION
# ---------------------------------------------------------

print(f"Catch time: {catch_time:.1f} minutes")
print(f"Catch distance: {catch_distance:.1f} km")


# ---------------------------------------------------------
# 5. CREATE THE GRAPH
# ---------------------------------------------------------

fig, ax = plt.subplots(figsize=(10, 6))

# Plot the robber's distance.
ax.plot(
    time,
    robber_distance,
    color='green',
    label='Robber: d = 2.5t'
)

# Plot the sheriff's distance.
ax.plot(
    time,
    sheriff_distance,
    color='brown',
    label='Sheriff: d = 3(t - 5)'
)


# ---------------------------------------------------------
# 6. MARK THE CAPTURE POINT
# ---------------------------------------------------------

# Mark the point where the two lines intersect.
ax.scatter(
    catch_time,
    catch_distance,
    color='purple',
    s=80,
    zorder=5,
    label=f'Capture: ({catch_time:.0f} min, {catch_distance:.0f} km)'
)

# Draw a vertical line showing the capture time.
ax.axvline(
    x=catch_time,
    color='purple',
    linestyle='--',
    alpha=0.6
)

# Draw a horizontal line showing the capture distance.
ax.axhline(
    y=catch_distance,
    color='purple',
    linestyle='--',
    alpha=0.6
)


# ---------------------------------------------------------
# 7. FORMAT THE GRAPH
# ---------------------------------------------------------

ax.set_title('A Bank Robber Caught', fontsize=14)

ax.set_xlabel('Time (minutes)')
ax.set_ylabel('Distance (km)')

ax.set_xlim(0, 40)
ax.set_ylim(0, 100)

ax.grid(True, alpha=0.3)
ax.legend()

plt.show()
````

````python
import sympy as sp
import matplotlib.pyplot as plt

# 1. DEFINE THE VARIABLES
# t = number of days after April 1st
t = sp.symbols('t')

# 2. DEFINE THE ENERGY EQUATIONS
# Mark I starts generating 1 kJ/day on April 1st.
# Therefore, after t days:
#     E_I = 1 * t
# Since the rate is 1 kJ/day, the total energy is simply t kJ.

mark_I = 1 * t

# Mark II starts generating energy on May 1st.
# April has 30 days, so Mark II starts 30 days after April 1st.
# Mark II generates 4 kJ/day.
# Therefore:
#     E_II = 4 * (t - 30)
# This equation is valid only after day 30.

mark_II = 4 * (t - 30)

# 3. FIND WHEN BOTH DESIGNS HAVE GENERATED THE SAME TOTAL ENERGY
# They have generated the same amount when:
#     Mark I energy = Mark II energy
#     t = 4(t - 30)

equation = sp.Eq(mark_I, mark_II)

# Ask SymPy to solve the equation for t.
solution = sp.solve(equation, t)

print("Solution for t:", solution)

# 4. CALCULATE THE TOTAL ENERGY

# Extract the solution.
catch_day = solution[0]

# Calculate the energy generated by Mark I at the
# moment when both designs have generated the same amount.
energy_I = mark_I.subs(t, catch_day)

# Calculate the energy generated by Mark II at the
# same moment.
energy_II = mark_II.subs(t, catch_day)

# Add the energy generated by both designs.
total_energy = energy_I + energy_II

print(f"Day after April 1st: {catch_day}")
print(f"Total energy generated: {total_energy} kJ")

# 5. CONVERT THE DAY INTO A CALENDAR DATE

# April 1st + 40 days = May 11th.
# We can use Python's datetime module to calculate this
# automatically instead of manually determining the date.

from datetime import date, timedelta

start_date = date(2026, 4, 1)

capture_date = start_date + timedelta(days=int(catch_day))

print(f"Calendar date: {capture_date.strftime('%B %d, %Y')}")


# 6. QUESTION 3

# What happens if Mark II also generates only 1 kJ/day?
# Then:
#     Mark I = t
#     Mark II = t - 30
# Mark II starts 30 days later but has exactly the same
# production rate.
# Therefore, it can NEVER catch up with Mark I.

mark_II_slow = 1 * (t - 30)

equation_question_3 = sp.Eq(mark_I, mark_II_slow)

solution_question_3 = sp.solve(equation_question_3, t)

print("\nQuestion 3:")
print("Equation:", equation_question_3)
print("Solution:", solution_question_3)

if not solution_question_3:
    print("Mark II never catches up with Mark I.")
    print("Mark I always has a 30 kJ advantage.")

# 7. VISUALIZE THE PROBLEM

# Create numerical values for the graph.
days = list(range(0, 80))

# Calculate cumulative energy for Mark I.
energy_I = [1 * day for day in days]

# Calculate cumulative energy for Mark II.
# Before day 30, Mark II has generated 0 kJ.
energy_II = [
    0 if day < 30 else 4 * (day - 30)
    for day in days
]


# Create the graph.
fig, ax = plt.subplots(figsize=(10, 6))

# Plot Mark I.
ax.plot(
    days,
    energy_I,
    label='Mark I: 1 kJ/day'
)

# Plot Mark II.
ax.plot(
    days,
    energy_II,
    label='Mark II: 4 kJ/day'
)


# Mark the point where they have generated
# the same total energy.
ax.scatter(
    int(catch_day),
    float(total_energy),
    s=80,
    zorder=5,
    label=f'Equal energy: day {catch_day}'
)


# Add labels.
ax.set_title('Cumulative Energy Generated')
ax.set_xlabel('Days after April 1st')
ax.set_ylabel('Total Energy (kJ)')

# Add grid and legend.
ax.grid(True, alpha=0.3)
ax.legend()

plt.show()
````

## Topic 04: Scalars

### Scalar

A **scalar** is a single numerical value. Unlike a vector or matrix, a scalar has only one component.

For example:

$$
c = 5
$$

Scalars are commonly used to **scale vectors or matrices**. Every element is multiplied by the scalar.

For example:

$$
4
\begin{bmatrix}
2 \\
3
\end{bmatrix}
=
\begin{bmatrix}
8 \\
12
\end{bmatrix}
$$

Here, $4$ is the **scalar**, and it multiplies every component of the vector.

**Key idea:** A scalar is simply a single number used to represent a magnitude or to scale other mathematical objects. numbers ($\mathbb{C}$), depending on the mathematical space being used.

## 06 - Norms and Unit Vectors

The **L2 norm** measures the length (or magnitude) of a vector. It represents the Euclidean distance from the origin to the point represented by the vector.

For a vector $\mathbf{v} = [v_1, v_2, \dots, v_n]$, the L2 norm is:

$$
\|\mathbf{v}\|_2 = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2}
$$

For example:

$$
\mathbf{v} = [3,4]
\quad\Rightarrow\quad
\|\mathbf{v}\|_2 = \sqrt{3^2 + 4^2} = 5
$$

In Python:

```python
np.linalg.norm(v)

````python
import numpy as np
import matplotlib.pyplot as plt

# Define a vector
v = np.array([3, 4])

# Calculate the L2 norm
l2_norm = np.linalg.norm(v)

print(f"Vector: {v}")
print(f"L2 norm: {l2_norm}")

# Plot the vector
fig, ax = plt.subplots(figsize=(6, 6))

ax.quiver(
    0, 0,
    v[0], v[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='blue'
)

# Draw the L2 norm as a dashed line
ax.plot(
    [0, v[0]],
    [0, v[1]],
    '--',
    color='red',
    label=f'L2 norm = {l2_norm:.1f}'
)

# Mark the endpoint
ax.scatter(v[0], v[1], color='black', zorder=3)

ax.set_xlim(-1, 5)
ax.set_ylim(-1, 5)

ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_title(r'L2 Norm of $\mathbf{v} = [3,4]$')

ax.grid(True, linestyle='--', alpha=0.5)
ax.set_aspect('equal')
ax.legend()

plt.show()
````

A **unit vector** is a vector with a **magnitude (L2 norm) equal to 1**. It represents a direction without changing its scale.

For a non-zero vector $\mathbf{v}$, its corresponding unit vector is obtained by dividing the vector by its L2 norm:

$$
\hat{\mathbf{v}} = \frac{\mathbf{v}}{\|\mathbf{v}\|_2}
$$

For example:

$$
\mathbf{v} = [3,4]
$$

Since $\|\mathbf{v}\|_2 = 5$:

$$
\hat{\mathbf{v}} = \frac{1}{5}[3,4]
= [0.6,0.8]
$$

and:

$$
\|\hat{\mathbf{v}}\|_2 = 1
$$

**Important:** A unit vector preserves the **direction** of the original vector but has a magnitude of exactly **1**.

## 07 - Basis, Orthogonal and Orthonormal Vectors

A **basis** is a set of vectors that is **linearly independent** and can **span** a vector space. In $\mathbb{R}^n$, a basis contains exactly $n$ linearly independent vectors.

Two vectors are **orthogonal** if they are perpendicular. This occurs when their **dot product is zero**:

$$
\mathbf{u} \cdot \mathbf{v} = 0
$$

A set of vectors is **orthogonal** if every pair of distinct vectors in the set is orthogonal.

For example, the standard basis of $\mathbb{R}^2$ is:

$$
\mathbf{e}_1 = [1,0], \qquad
\mathbf{e}_2 = [0,1]
$$

These vectors are orthogonal because:

$$
\mathbf{e}_1 \cdot \mathbf{e}_2
= (1)(0) + (0)(1) = 0
$$

Therefore, $\{\mathbf{e}_1,\mathbf{e}_2\}$ is an **orthogonal basis** for $\mathbb{R}^2$.

If every vector in an orthogonal basis also has magnitude $1$, it is called an **orthonormal basis**.

````python
import numpy as np
import matplotlib.pyplot as plt

# Standard basis vectors
e1 = np.array([1, 0])
e2 = np.array([0, 1])

# Create a new vector using the basis vectors
a = 2
b = 1

w = a * e1 + b * e2

fig, ax = plt.subplots(figsize=(7, 7))

# Plot e1
ax.quiver(
    0, 0, e1[0], e1[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='blue',
    label=r'$\mathbf{e}_1 = [1,0]$'
)

# Plot e2
ax.quiver(
    0, 0, e2[0], e2[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='red',
    label=r'$\mathbf{e}_2 = [0,1]$'
)

# Plot the new vector w
ax.quiver(
    0, 0, w[0], w[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='green',
    linewidth=2.5,
    label=r'$\mathbf{w} = 2\mathbf{e}_1 + \mathbf{e}_2 = [2,1]$'
)

# Coordinate axes
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(0, color='black', linewidth=0.8)

# Grid
ax.grid(True, linestyle='--', alpha=0.5)

# Limits
ax.set_xlim(-1, 3)
ax.set_ylim(-1, 3)

# Labels
ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_title('Basis Vectors and Linear Combination')

ax.set_aspect('equal')
ax.legend()

plt.show()
````
