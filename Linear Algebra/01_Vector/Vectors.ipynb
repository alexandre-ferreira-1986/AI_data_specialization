<h1 style="color:#1565C0;">04 - Vector-vector multiplication: the dot product</h1>

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
## Many ways to compute the dot product

# Create the first vector with 6 elements
v1 = np.array([ 1, 2, 3, 4, 5])

# Create the second vector with 5 elements
# NOTE: For a valid dot product, both vectors must have the same number of elements
v2 = np.array([ 0, -4,  -3, 6, 5 ])

# Method 1
# np.multiply() performs element-wise multiplication:
# [v1[0]*v2[0], v1[1]*v2[1], ..., v1[n]*v2[n]]
# sum() then adds all the resulting products together
dp1 = sum( np.multiply(v1,v2) )

# Method 2
# np.dot() directly calculates the dot product between the two vectors
# This is the **most common** and concise NumPy method for computing a dot product
dp2 = np.dot( v1,v2 )

# Method 3
# np.matmul() performs matrix multiplication
# When both inputs are 1-dimensional arrays, it produces the same result as np.dot()
dp3 = np.matmul( v1,v2 )

# Method 4
# Manually calculate the dot product
# Start with zero and accumulate the products of corresponding elements
dp4 = 0  # initialize

# Loop over all elements of the vectors
# For each position, multiply the corresponding elements
# and add the result to the accumulated dot product
for i in range(len(v1)):
    
    # Multiply the corresponding elements of v1 and v2
    # and add the result to the running total
    dp4 = dp4 + v1[i]*v2[i]


# Display the results obtained using the four different methods
# All methods should produce the same dot product value
print(dp1,dp2,dp3,dp4)
```

```python
# Define two vectors in 3D space
v1 = np.array([ 2,  4, -3 ])
v2 = np.array([ 0, -3, -3 ])

# Compute the angle between the two vectors in radians.
# The angle is calculated using the dot product formula:
# cos(theta) = (v1 · v2) / (||v1|| ||v2||)
# np.dot() computes the dot product, while np.linalg.norm()
# computes the magnitude (length) of each vector.
ang = np.arccos(
    np.dot(v1, v2) /
    (np.linalg.norm(v1) * np.linalg.norm(v2)))

# Create a figure for the 3D visualization
fig = plt.figure()

# Create a 3D coordinate system (axes)
ax = plt.axes(projection='3d')

# Draw the first vector as a blue line starting at the origin
# and ending at the coordinates defined by v1.
ax.plot(
    [0, v1[0]],
    [0, v1[1]],
    [0, v1[2]],
    'b')

# Draw the second vector as a red line starting at the origin
# and ending at the coordinates defined by v2.
ax.plot(
    [0, v2[0]],
    [0, v2[1]],
    [0, v2[2]],
    'r')

# Set the same limits for all three axes to make the visualization
# easier to interpret and preserve the proportions of the vectors.
plt.axis((-6, 6, -6, 6, -6, 6))

# Display the calculated angle in radians as the plot title
plt.title('Angle between vectors: %s rad.' % ang)

# Display the 3D plot
plt.show()
```

```python
## equivalence of algebraic and geometric dot product formulas

# two vectors
v1 = np.array([ 2,  4, -3 ])
v2 = np.array([ 0, -3, -3 ])


# algebraic
dp_a = np.dot( v1,v2 )

# geometric
dp_g = np.linalg.norm(v1)*np.linalg.norm(v2)*np.cos(ang)

# print dot product to command
print(dp_a)
print(dp_g)
```

<h1 style="color:#1565C0;">09 - Vector outer product</h1>

The outer product of two vectors produces a matrix by multiplying every element of the first vector by every element of the second vector.

For vectors \(\mathbf{u} \in \mathbb{R}^m\) and \(\mathbf{v} \in \mathbb{R}^n\):

$$ \mathbf{u}\mathbf{v}^T = \begin{bmatrix} u_1v_1 & u_1v_2 & \cdots & u_1v_n\\ u_2v_1 & u_2v_2 & \cdots & u_2v_n\\ \vdots & \vdots & \ddots & \vdots\\ u_mv_1 & u_mv_2 & \cdots & u_mv_n \end{bmatrix} $$

Key idea:

The outer product combines every element of one vector with every element of the other, resulting in an \(m x n\) matrix.

For example, let:

$$ \mathbf{u} = \begin{bmatrix} 1\\ 2\\ 3 \end{bmatrix}, \qquad \mathbf{v} = \begin{bmatrix} 4\\ 5 \end{bmatrix} $$

The outer product is:

$$ \mathbf{u}\mathbf{v}^T = \begin{bmatrix} 1\\ 2\\ 3 \end{bmatrix} \begin{bmatrix} 4 & 5 \end{bmatrix} = \begin{bmatrix} 1\cdot4 & 1\cdot5\\ 2\cdot4 & 2\cdot5\\ 3\cdot4 & 3\cdot5 \end{bmatrix} = \begin{bmatrix} 4 & 5\\ 8 & 10\\ 12 & 15 \end{bmatrix} $$

Thus, the result is a 3x2 matrix.

```python
import numpy as np

# two vectors
v1 = np.array([  1, 2, 3 ])
v2 = np.array([ -1, 0, 1 ])

# outer product
print(np.outer(v1,v2))
print("\n-------------\n")

# terrible programming, but helps conceptually:
op = np.zeros((len(v1),len(v2)))
for i in range(0,len(v1)):
    for j in range(0,len(v2)):
        op[i,j] = v1[i] * v2[j]

print(op)
```

<h1 style="color:#1565C0;">10 - Vector cross product</h1>

The cross product is an operation between two vectors in 3D space that **produces a new vector perpendicular to both input vectors.**

For two vectors **u** and **v**:

$$ \mathbf{u}\times\mathbf{v} = \begin{bmatrix} u_2v_3-u_3v_2\\ u_3v_1-u_1v_3\\ u_1v_2-u_2v_1 \end{bmatrix} $$
Example

Let:

$$ \mathbf{u} = \begin{bmatrix} 1\\ 2\\ 3 \end{bmatrix}, \qquad \mathbf{v} = \begin{bmatrix} 4\\ 5\\ 6 \end{bmatrix} $$

Then:

$$ \mathbf{u}\times\mathbf{v} = \begin{bmatrix} 2(6)-3(5)\\ 3(4)-1(6)\\ 1(5)-2(4) \end{bmatrix} = \begin{bmatrix} 12-15\\ 12-6\\ 5-8 \end{bmatrix} = \begin{bmatrix} -3\\ 6\\ -3 \end{bmatrix} $$

So:

$$ \boxed{ \mathbf{u}\times\mathbf{v} = \begin{bmatrix} -3\\ 6\\ -3 \end{bmatrix} } $$

Key idea: The cross product returns a vector perpendicular to both \(\mathbf{u}\) and \(\mathbf{v}\).

```python
import numpy as np
import matplotlib.pyplot as plt

# create vectors
v1  = [ -3,  2, 5 ]
v2  = [  4, -3, 0 ]

# Python's cross-product function
v3a = np.cross( v1,v2 )

# "manual" method
v3b = [ [v1[1]*v2[2] - v1[2]*v2[1]],
        [v1[2]*v2[0] - v1[0]*v2[2]],
        [v1[0]*v2[1] - v1[1]*v2[0]] ]

print(v3a,v3b)


fig = plt.figure()
# ax = fig.add_subplot(projection='3d')
ax = plt.axes(projection = '3d')

# draw plane defined by span of v1 and v2
xx, yy = np.meshgrid(np.linspace(-10,10,10),np.linspace(-10,10,10))
z1 = (-v3a[0]*xx - v3a[1]*yy)/v3a[2]
ax.plot_surface(xx,yy,z1,alpha=.2)

## plot the two vectors
ax.plot([0, v1[0]],[0, v1[1]],[0, v1[2]],'k')
ax.plot([0, v2[0]],[0, v2[1]],[0, v2[2]],'k')
ax.plot([0, v3a[0]],[0, v3a[1]],[0, v3a[2]],'r')


ax.view_init(azim=150,elev=45)
plt.show()
```

<h1 style="color:#1565C0;">11 - Hermitian transpose (a.k.a. conjugate transpose)</h1>

The Hermitian transpose of a vector or matrix is obtained by transposing it and taking the complex conjugate of each element.

It is usually denoted by:

$$ A^H $$

If \(A\) contains complex numbers:

$$ A^H = \overline{A}^{\,T} $$
Example

Consider the complex vector:

$$ \mathbf{v} = \begin{bmatrix} 1+i\\ 2-3i \end{bmatrix} $$

First, take the complex conjugate:

$$ \overline{\mathbf{v}} = \begin{bmatrix} 1-i\\ 2+3i \end{bmatrix} $$

Then transpose it:

$$ \mathbf{v}^H = \begin{bmatrix} 1-i & 2+3i \end{bmatrix} $$

So:

$$ \boxed{ \mathbf{v}^H = \begin{bmatrix} 1-i & 2+3i \end{bmatrix} } $$
**Why is it important?**

The Hermitian transpose is essential when working with complex vectors and matrices, particularly in:

Quantum mechanics
Signal processing
Fourier analysis
Complex-valued linear algebra
Machine learning involving complex numbers

For example, the inner product of a complex vector is defined using the Hermitian transpose:

$$ \mathbf{v}^H\mathbf{v} $$

For the vector above:

$$ \mathbf{v}^H\mathbf{v} = (1-i)(1+i)+(2+3i)(2-3i) $$ $$ =2+13=15 $$

Notice that the result is a real, non-negative number, which is an important property of complex inner products.

**Key idea:**

*The Hermitian transpose is the complex-number equivalent of the ordinary transpose, with complex conjugation added to preserve the mathematical properties of inner products.*

```python
import numpy as np

# create a complex number
z = complex(3,4)
print(z)

# magnitude
print( np.linalg.norm(z) )

# by transpose?
print( np.transpose(z)*z )

# by Hermitian transpose
print( np.transpose(z.conjugate())*z )


# complex vector
v = np.array( [ 3, 4j, 5+2j, complex(2,-5) ] )
print( v.T )
print( np.transpose(v) )
print( np.transpose(v.conjugate()) )
```

<h1 style="color:#1565C0;">12 - Unit vector</h1>



A **unit vector** is a vector whose **magnitude (length) is exactly 1**.

Formally, a nonzero vector $\mathbf{v}$ is a unit vector if:

$$
\boxed{\|\mathbf{v}\| = 1}
$$

Given any **nonzero** vector $\mathbf{v}$, we can create a unit vector in the **same direction** by dividing the vector by its magnitude:

$$
\boxed{
\hat{\mathbf{v}} = \frac{\mathbf{v}}{\|\mathbf{v}\|}
}
$$

where:

* $\mathbf{v}$ = original vector
* $|\mathbf{v}|$ = magnitude (length) of the vector
* $\hat{\mathbf{v}}$ = unit vector in the same direction

## Example

Consider the vector:

$$
\mathbf{v} =
\begin{bmatrix}
3\\
4
\end{bmatrix}
$$

Its magnitude is:

$$
\|\mathbf{v}\|
=
\sqrt{3^2+4^2}
=
5
$$

Therefore, the unit vector in the same direction is:

$$
\hat{\mathbf{v}}
=
\frac{\mathbf{v}}{\|\mathbf{v}\|}
=
\frac{1}{5}
\begin{bmatrix}
3\\
4
\end{bmatrix}
=
\begin{bmatrix}
\frac{3}{5}\\
\frac{4}{5}
\end{bmatrix}
$$

We can verify that its magnitude is exactly $1$:

$$
\left\|\hat{\mathbf{v}}\right\|
=
\sqrt{
\left(\frac{3}{5}\right)^2+
\left(\frac{4}{5}\right)^2
}
=
\sqrt{
\frac{9}{25}+\frac{16}{25}
}
=
\sqrt{1}
=
1
$$

## Important Exception: The Zero Vector

The zero vector cannot be normalized:

$$
\mathbf{0} =
\begin{bmatrix}
0\\
0
\end{bmatrix}
$$

because:

$$
\|\mathbf{0}\| = 0
$$

Therefore:

$$
\frac{\mathbf{0}}{\|\mathbf{0}\|}
=
\frac{\mathbf{0}}{0}
$$

is **undefined**.

Geometrically, this makes sense because the zero vector has **no direction**. Therefore, there is no unit vector pointing in the same direction as the zero vector.

### Key Idea

> A unit vector has length $1$. Any **nonzero** vector can be converted into a unit vector by dividing it by its magnitude:

$$
\boxed{
\hat{\mathbf{v}} = \frac{\mathbf{v}}{\|\mathbf{v}\|}
}
$$

```python
import numpy as np

# Define two vectors
a = np.random.randn(3)
b = np.random.randn(3)

# Compute the dot product a^T b
dot_product = np.dot(a, b)

# Compute the magnitude (Euclidean norm) of each vector
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

# Cauchy-Schwarz inequality:
#
# |a^T b| <= ||a|| ||b||
#
# Left-hand side
left_side = abs(dot_product)

# Right-hand side
right_side = norm_a * norm_b


# Display the values
print(f"|a^T b|     = {left_side:.4f}")
print(f"||a|| ||b|| = {right_side:.4f}")


# Verify the Cauchy-Schwarz inequality
print("\nCauchy-Schwarz inequality satisfied:", left_side <= right_side)
```

```python
import numpy as np

# ---------------------------------------------------------
# Create two random vectors
# ---------------------------------------------------------

n = 4  # Number of elements (dimensions) in each vector

# Generate random values from a standard normal distribution,
# scale them by 20, and round them to integers.
v1 = np.round(20 * np.random.randn(n))
v2 = np.round(20 * np.random.randn(n))


# ---------------------------------------------------------
# Compute vector magnitudes and the magnitude of the dot product
# ---------------------------------------------------------

# The magnitude (Euclidean length) of a vector is:
#
#       ||v|| = sqrt(v · v)
#
# np.dot(v1, v1) computes the dot product of v1 with itself.
# Taking the square root gives the magnitude of v1.

v1m = np.sqrt(np.dot(v1, v1))

# Same calculation for v2
v2m = np.sqrt(np.dot(v2, v2))

# Compute the dot product between v1 and v2.
#
# The absolute value is used here because we are interested
# only in the magnitude of the dot product, not its sign.
#
#       |v1 · v2|

dpm = np.abs(np.dot(v1, v2))

print(v1m, v2m, dpm)
print(' ')


# ---------------------------------------------------------
# Normalize the vectors
# ---------------------------------------------------------

# A unit vector is obtained by dividing a vector by its magnitude:
#
#       v_hat = v / ||v||
#
# After normalization, the vector has length 1 while
# maintaining the same direction as the original vector.

v1u = v1 / v1m
v2u = v2 / v2m

print(v1u, v2u)
print(' ')


# ---------------------------------------------------------
# Compute the dot product of the unit vectors
# ---------------------------------------------------------

# Calculate the dot product of the normalized vectors:
#
#       |v1_hat · v2_hat|
#
# Since both vectors now have magnitude 1, the dot product
# is directly related to the cosine of the angle between them:
#
#       v1_hat · v2_hat = cos(theta)
#
# Therefore, the result will always be between 0 and 1
# because we are taking the absolute value.

dpm = np.abs(np.dot(v1u, v2u))

print(dpm)
```
